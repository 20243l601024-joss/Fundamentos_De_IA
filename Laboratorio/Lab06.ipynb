```python
#Importar las librerias
#librerias para graficar
import matplotlib.pyplot as plt 
#librerias que permite trabajar con estructuras complejas, como Grafos
import networkx as nx 
```

```python
#creamos la instancia
grafo = nx.Graph()
```

```python
#Creamos el grafo utilizando el metodo add_edges
grafo.add_edges_from(
[
    ("A","B"),
    ("A","C"),
    ("C","G"),
    ("D","F"),
    ("F","H"),
    ("H","I"),
    ("I","J"),
    ("F","E"),
    ("G","E"),
    ("G","J")
]
     )
```

```python
nx.draw(
    grafo,
    with_labels =True,
    node_size=1000,
    node_color="#e44b8d",
    node_shape="o"
)
plt.show()
```

```python
rutas = list(
    nx.all_simple_paths(grafo,source="A", target="J")
)
```

```python
print("Rutas encontradas:")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}:{ruta}")
```

```python
#mostrar el total de pasos de cada ruta
pasos1 = len(rutas[0])-1
print("Pasos de la ruta 1:", pasos1)
pasos2 = len(rutas[1])-1
print("Pasos de la ruta 2:", pasos2)
```

```python
P = pasos1 < pasos2
Q = pasos2 < pasos1 

if P==True:
    print("P = True")
    print("La ruta 1 es mejor opcion.")
elif Q==True:
    print("Q = True")
    print("La ruta 2 es mejor opcion.")

else:
    print("P = False")
    print("Q = False")
    print("Las cuatro rutas son igual de buenas, haz lo que quieras!")
```
