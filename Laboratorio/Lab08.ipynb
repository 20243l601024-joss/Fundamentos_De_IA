```python
#Importar las librerias
#librerias para graficar
import matplotlib.pyplot as plt 
#librerias que permite trabajar con estructuras complejas, como Grafos
import networkx as nx 
```

```python
grafoServidor = nx.Graph()
```

```python
#Creamos el grafo utilizando el metodo add_edges
grafoServidor.add_edges_from(
[
    ("Servidor_1","RouterA"),
    ("RouterA","SwitchA"),
    ("RouterA","SwitchB"),
    ("SwitchA","ImpresoraA"),
    ("SwitchA","PC1"),
    ("SwitchB","PC2"),
    ("SwitchB","ImpresoraA")
]       )
```

```python
nx.draw(
    grafoServidor,
    with_labels =True,
    node_size=1000,
    node_color="#e44b8d",
    node_shape="o"
)
plt.show()
```
