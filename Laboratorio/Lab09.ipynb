```python
#Importar las librerias
#librerias para graficar
import matplotlib.pyplot as plt 
#librerias que permite trabajar con estructuras complejas, como Grafos
import networkx as nx
```

```python
grafo = nx.DiGraph()
grafo.add_weighted_edges_from(
    [
        ("A", "B",5),
        ("A", "C",2),
        ("B", "D",4),
        ("C", "E",5),
        ("C", "F",6)
    ])
```

```python
#Imprimimos el grafo
pos = nx.spring_layout( grafo, seed=10)
nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)
labels =nx.get_edge_attributes (grafo, "weight")
nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
#puedo moverme de "A a B"
print(grafo.has_edge("A", "B"))
```

```python
#puedo moverme de "B a A"
print(grafo.has_edge("B", "A"))
#puedo moverme de "A a F"
print(grafo.has_edge("A", "F"))
```

```python
#Mostramos las rutas
rutas=list(
    nx.all_simple_paths(
        grafo,
        source="A",
        target="F"
    )
)
```

```python
#Imprimimos las rutas
for i, ruta in enumerate (rutas,start=1):
    print (f"Ruta {i}:{ruta}")
```

```python
grafo2 = nx.DiGraph()
grafo2.add_weighted_edges_from(
    [
        ("A", "B",4),
        ("A", "C",7),
        ("B", "D",3),
        ("B", "E",6),
        ("C", "D",2),
        ("C", "F",5),
        ("D", "E",2),
        ("D", "G",8),
        ("F", "G",4),
        ("E", "G",3)
    ])
```

```python
pos = nx.spring_layout( grafo2, seed=12)
nx.draw(
    grafo2,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)
labels =nx.get_edge_attributes (grafo2, "weight")
nx.draw_networkx_edge_labels(
    grafo2,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
#Mostramos las rutas
rutas=list (
    nx.all_simple_paths(
        grafo2,
        source="A",
        target="G"
    )
)
```

```python
#Imprimimos las rutas
for i, ruta in enumerate (rutas,start=1):
    print (f"Ruta {i}:{ruta}")
```

```python
#calculamos el costo
def costo_ruta (grafo2,ruta):
    costo=0
    for i in range (len(ruta) -1):
        costo += grafo2 [ruta[i]] [ruta[i+1]] ["weight"]
    return costo
```

```python
#Imprimimos el costo
for i, ruta in enumerate (rutas, start=1):
    costo=costo_ruta(grafo2,ruta)
    print (f"Ruta{i}: {ruta}-> {costo} unidades")
```

```python
##Elegir la ruta mas corta
costos=[ costo_ruta (grafo2,ruta)
         for ruta in rutas
        ]
```

```python
mejor_ruta= costos.index(min(costos))
```

```python
##Imprimit la mejor ruta
print (f"La mejor ruta es:{mejor_ruta + 1}")
print (f"El camino corto es:{costos[mejor_ruta]}")
```
