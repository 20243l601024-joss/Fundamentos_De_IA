```python
servidor= input("El servidor esta disponible?(s/n):").lower() == "n"
latencia = input("La latencia es alta?(s/n):").lower() =="s"
paquetesperdidos =input("Existe perdida de paquetes? (s/n):").lower() == "s"
trafico =input("El trafico es alto?(s/n):").lower() == "s"
respaldoServidor =input("El servidor de respaldo esta disponible? (s/n):").lower() =="n"
SistemaMonitoreo=input("El sistema de monitoreo esta disponible? (s/n):").lower() == "s"
Servicio=(servidor and latencia and paquetesperdidos and trafico and respaldoServidor and SistemaMonitoreo)
if Servicio == True:
    print ("El servicio presenta condiciones criticas ")
else:
    print("El servicio no presenta una condicion critica")
```
